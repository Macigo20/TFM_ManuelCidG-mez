```python
import pandas as pd

df = pd.read_csv("banking_conversations.csv")
unique_ids = df["conversation_id"].unique()
sample_ids = pd.Series(unique_ids).sample(n=3000, random_state=42)

sample_df = df[df["conversation_id"].isin(sample_ids)]
sample_df.to_csv("muestra_3000_conversaciones.csv", index=False)

print(f"{sample_df.shape[0]} filas, {sample_df['conversation_id'].nunique()} conversaciones")
```
